```python
df = spark.table("science_home.`db-inorganics`.mp_gold")
#display(df.dtypes)
#display(df.select("id", "formula"))
```

```python
# Drop all string and boolean columns except 'id'
string_cols = [name for name, dtype in df.dtypes if dtype == 'string' and name != 'id']
bool_cols = [name for name, dtype in df.dtypes if dtype == 'boolean']
extra_cols = ['types_of_magnetic_species', 'possible_species', 'elements']
cols_to_drop = string_cols + bool_cols + extra_cols

df_dielectric_dropped = df.drop(*cols_to_drop)

len(df_dielectric_dropped.columns)
```

```python
null_cols_to_drop = ['types_of_magnetic_species', 'possible_species', 'elements', 'weighted_work_function', 'weighted_surface_energy', 'weighted_surface_energy_ev_per_ang_2', 'surface_energy', 'surface_anisotropy', 'shape_factor', 'e_ij_max', 'e_ionic', 'e_electronic']

df_dielectric_null_dropped = df_dielectric_dropped.drop(*null_cols_to_drop)
len(df_dielectric_null_dropped.columns)

```

```python
df_dielectric_null_dropped.write.mode("overwrite").format("delta").saveAsTable("science_home.`ml-dielectric`.dielectric_silver")
```
