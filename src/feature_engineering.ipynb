````python
from pyspark.sql import SparkSession
# Ensure Spark session is active
spark = SparkSession.builder.getOrCreate()

# Read dielectric_silver table from Unity Catalog
df_silver = spark.table("science_home.`ml-dielectric`.dielectric_silver")

# Read mp_gold table to get formula and structure columns for feature engineering
df_gold = spark.table("science_home.`db-inorganics`.mp_gold").select("id", "formula", "structure")

# Join to get formula and structure alongside dielectric properties
df_joined = df_silver.join(df_gold, on="id", how="inner")

print(f"Total rows: {df_joined.count()}")
print(f"Columns: {len(df_joined.columns)}")
display(df_joined.select("id", "formula", "structure", "dielectric_constant", "band_gap", "density").limit(5))
````

````python
import pandas as pd
import numpy as np
from pymatgen.core import Composition, Structure
from matminer.featurizers.composition import (
    ElementProperty,
    Stoichiometry,
    ValenceOrbital,
    BandCenter,
    OxidationStates,
    AtomicOrbitals,
    TMetalFraction,
)
from matminer.featurizers.structure import (
    DensityFeatures,
    OrbitalFieldMatrix,
    MaximumPackingEfficiency,
)

print("All matminer featurizers imported successfully.")
````

````python
import json

# Convert Spark DataFrame to Pandas for matminer processing
df_pd = df_joined.toPandas()

# Parse formula strings into pymatgen Composition objects
df_pd["composition"] = df_pd["formula"].apply(lambda x: Composition(x) if pd.notna(x) else None)

# Parse structure strings into pymatgen Structure objects
def parse_structure(s):
    """Parse structure JSON string into pymatgen Structure."""
    try:
        if pd.notna(s):
            d = json.loads(s)
            return Structure.from_dict(d)
    except Exception:
        return None
    return None

df_pd["structure_obj"] = df_pd["structure"].apply(parse_structure)

# Check how many structures were successfully parsed
n_comp = df_pd["composition"].notna().sum()
n_struct = df_pd["structure_obj"].notna().sum()
print(f"Successfully parsed compositions: {n_comp}/{len(df_pd)}")
print(f"Successfully parsed structures: {n_struct}/{len(df_pd)}")
print(f"\nDataFrame shape: {df_pd.shape}")
````

````python
# ============================================================
# COMPOSITION FEATURES
# ============================================================

# 1. ElementProperty (Magpie) - statistics of elemental properties
ep_feat = ElementProperty.from_preset(preset_name="magpie")
ep_feat.set_n_jobs(1)
df_pd = ep_feat.featurize_dataframe(df_pd, col_id="composition", ignore_errors=True)
print(f"After ElementProperty (Magpie): {df_pd.shape[1]} columns")

# 2. Stoichiometry - stoichiometric attributes
stoich_feat = Stoichiometry()
stoich_feat.set_n_jobs(1)
df_pd = stoich_feat.featurize_dataframe(df_pd, col_id="composition", ignore_errors=True)
print(f"After Stoichiometry: {df_pd.shape[1]} columns")

# 3. ValenceOrbital - valence orbital occupation statistics
vo_feat = ValenceOrbital()
vo_feat.set_n_jobs(1)
df_pd = vo_feat.featurize_dataframe(df_pd, col_id="composition", ignore_errors=True)
print(f"After ValenceOrbital: {df_pd.shape[1]} columns")

# 4. BandCenter - estimate of band center
bc_feat = BandCenter()
bc_feat.set_n_jobs(1)
df_pd = bc_feat.featurize_dataframe(df_pd, col_id="composition", ignore_errors=True)
print(f"After BandCenter: {df_pd.shape[1]} columns")

# 5. OxidationStates - oxidation state statistics
os_feat = OxidationStates()
os_feat.set_n_jobs(1)
df_pd = os_feat.featurize_dataframe(df_pd, col_id="composition", ignore_errors=True)
print(f"After OxidationStates: {df_pd.shape[1]} columns")

# 6. AtomicOrbitals - highest occupied / lowest unoccupied orbitals
ao_feat = AtomicOrbitals()
ao_feat.set_n_jobs(1)
df_pd = ao_feat.featurize_dataframe(df_pd, col_id="composition", ignore_errors=True)
print(f"After AtomicOrbitals: {df_pd.shape[1]} columns")

# 7. TMetalFraction - fraction of transition metals
tm_feat = TMetalFraction()
tm_feat.set_n_jobs(1)
df_pd = tm_feat.featurize_dataframe(df_pd, col_id="composition", ignore_errors=True)
print(f"After TMetalFraction: {df_pd.shape[1]} columns")

print(f"\n=== Total columns after composition features: {df_pd.shape[1]} ===")
````

````python
# ============================================================
# STRUCTURAL FEATURES
# ============================================================

# Filter to rows with valid structures for structural featurization
df_struct = df_pd[df_pd["structure_obj"].notna()].copy()
print(f"Rows with valid structures: {len(df_struct)}")

# Rename existing columns that conflict with featurizer output names
df_struct = df_struct.rename(columns={"density": "density_original"})

# 1. DensityFeatures - packing fraction, density
dens_feat = DensityFeatures()
dens_feat.set_n_jobs(1)
df_struct = dens_feat.featurize_dataframe(df_struct, col_id="structure_obj", ignore_errors=True)
print(f"After DensityFeatures: {df_struct.shape[1]} columns")
````

````python
# ============================================================
# FINALIZE FEATURES (Local environment features skipped - too expensive)
# ============================================================

# Drop intermediate object columns
df_features = df_struct.drop(columns=["composition", "structure_obj", "structure", "formula"])
print(f"Final features DataFrame: {df_features.shape}")
display(df_features.head())
````

````python
print('All experiments completed')
````

````python
# ============================================================
# SAVE FEATURIZED DATA TO UNITY CATALOG (Staging table)
# ============================================================
import re

# Sanitize column names: replace spaces and special characters with underscores
def sanitize_col_name(name):
    return re.sub(r'[\s;{}()\n\t=,]+', '_', name).strip('_')

df_features_clean = df_features.rename(columns={col: sanitize_col_name(col) for col in df_features.columns})

# Convert pandas DataFrame to Spark DataFrame for writing
df_features_spark = spark.createDataFrame(df_features_clean)

# Write as a staging table (will be promoted to Feature Table after pip install)
staging_table = "science_home.`ml-dielectric`.featurized_dielectric_staging"
df_features_spark.write.mode("overwrite").saveAsTable(staging_table)

print(f"Saved staging table: {staging_table}")
print(f"Rows: {df_features_spark.count()}")
print(f"Columns: {len(df_features_spark.columns)}")
````

## Publish Features to Unity Catalog Feature Store

**Why Feature Store?**
- Centralized, discoverable features (visible in Catalog Explorer)
- Primary key enforcement for point-in-time lookups
- Lineage tracking (which models use which features)
- Colleagues can consume via `fe.read_table()` or auto-join with `fe.create_training_set()`

**Steps:**
1. Import `FeatureEngineeringClient`
2. Drop the plain Delta table (cell 9 created it without feature metadata)
3. Recreate as a Feature Table with `id` as the primary key
4. Verify discoverability

````python
from databricks.feature_engineering import FeatureEngineeringClient
from pyspark.sql import SparkSession

spark = SparkSession.builder.getOrCreate()
fe = FeatureEngineeringClient()

# Read from the staging table (persisted by cell 9 across kernel restart)
df_features_spark = spark.table("science_home.`ml-dielectric`.featurized_dielectric_staging")
print("FeatureEngineeringClient initialized.")
print(f"Loaded from staging: {df_features_spark.count()} rows, {len(df_features_spark.columns)} columns")
````

````python
from databricks.feature_engineering import FeatureEngineeringClient
from pyspark.sql import SparkSession

spark = SparkSession.builder.getOrCreate()
fe = FeatureEngineeringClient()

# Read from the staging table (persisted across kernel restart)
df_features_spark = spark.table("science_home.`ml-dielectric`.featurized_dielectric_staging")
print("FeatureEngineeringClient initialized.")
print(f"Loaded from staging: {df_features_spark.count()} rows, {len(df_features_spark.columns)} columns")
````

````python
# Create the Feature Table in Unity Catalog
# This registers the table with primary key metadata for discovery, lineage, and auto-joins

table_name = "science_home.ml-dielectric.featurized_dielectric"

# Drop any existing table (plain or feature) at this location
spark.sql("DROP TABLE IF EXISTS science_home.`ml-dielectric`.featurized_dielectric")

# Create Feature Table with 'id' as primary key
fe.create_table(
    name=table_name,
    primary_keys=["id"],
    df=df_features_spark,
    description="Featurized dielectric materials dataset. Contains Magpie composition features, "
                "stoichiometry, valence orbital, band center, oxidation states, atomic orbitals, "
                "transition metal fraction, and structural density features. "
                "Primary key: Materials Project ID."
)

print(f"Feature table created: {table_name}")
print(f"Primary key: id")
````

````python
# Verify the feature table and clean up staging
df_verify = fe.read_table(name="science_home.ml-dielectric.featurized_dielectric")
print(f"Feature table verified: {df_verify.count()} rows x {len(df_verify.columns)} columns")

# Drop the staging table (no longer needed)
spark.sql("DROP TABLE IF EXISTS science_home.`ml-dielectric`.featurized_dielectric_staging")
print("Staging table dropped.")

display(df_verify.limit(5))
````
