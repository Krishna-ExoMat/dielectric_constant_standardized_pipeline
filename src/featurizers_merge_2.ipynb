```python
# Read all three sources silver data + BP featurizers + Matminer composition featurizers

df_silver = spark.table("science_home.`ml-dielectric`.dielectric_silver")
df_featurizer = spark.table("science_home.`ml-dielectric`.bp_dielectric_featurizer")
df_magpie = spark.table("science_home.`ml-dielectric`.compos_matminer_features")

print(f"df_silver: {df_silver.count()} rows, {len(df_silver.columns)} cols")
print(f"df_featurizer: {df_featurizer.count()} rows, {len(df_featurizer.columns)} cols")
print(f"df_magpie: {df_magpie.count()} rows, {len(df_magpie.columns)} cols")

# Merge all on 'id'
df_merged = df_featurizer.join(df_silver, on="id", how="inner")
df_final = df_merged.join(df_magpie.drop("formula"), on="id", how="inner")

print(f"\nFinal merged: {df_final.count()} rows, {len(df_final.columns)} cols")
```

```python
# Save to Unity Catalog
print("Saving to science_home.ml-dielectric.featurized_merged...")
df_final.write.mode("overwrite").saveAsTable("science_home.`ml-dielectric`.featurized_merged")
print(f" Saved successfully!")
```
